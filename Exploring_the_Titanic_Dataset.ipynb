```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình hiển thị biểu đồ đẹp hơn
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Đọc dữ liệu (đảm bảo file train.csv đã được upload lên Colab)
df = pd.read_csv('train.csv')

# Xem 5 dòng đầu tiên
df.head()
```

```python
# 1. Kích thước dữ liệu (Shape)
print("Shape của dữ liệu:", df.shape)
print("-" * 30)

# 2. Kiểu dữ liệu và thông tin chung
print(df.info())
print("-" * 30)

# 3. Kiểm tra dữ liệu bị thiếu (Missing values)
print("Số lượng giá trị thiếu ở mỗi cột:")
print(df.isnull().sum())
```

```python
# Tính tỷ lệ sống sót
survival_counts = df['Survived'].value_counts()
survival_rate = df['Survived'].mean()

print(f"Tỷ lệ sống sót chung: {survival_rate:.2%}")

# Vẽ biểu đồ
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Survived', palette='Set2')
plt.title('Phân bố sống sót (0 = Chết, 1 = Sống)')
plt.show()
```

```python
# Tính tần suất tương đối
gender_survival = df.groupby('Sex')['Survived'].value_counts(normalize=True).unstack()
print(gender_survival)

# Vẽ biểu đồ
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Sex', y='Survived', palette='pastel', errorbar=None)
plt.title('Tỷ lệ sống sót theo Giới tính')
plt.ylabel('Tỷ lệ sống sót')
plt.show()
```

```python
# Tính tần suất tương đối
class_survival = df.groupby('Pclass')['Survived'].value_counts(normalize=True).unstack()
print(class_survival)

# Vẽ biểu đồ
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Pclass', y='Survived', palette='viridis', errorbar=None)
plt.title('Tỷ lệ sống sót theo Hạng vé (Pclass)')
plt.ylabel('Tỷ lệ sống sót')
plt.xlabel('Hạng vé (1 = Cao cấp, 3 = Phổ thông)')
plt.show()
```

```python
# Tạo nhóm tuổi (Binning)
bins = [0, 12, 18, 35, 60, 100]
labels = ['Trẻ em', 'Thiếu niên', 'Thanh niên', 'Trung niên', 'Người già']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

# Tính tần suất tương đối
age_survival = df.groupby('AgeGroup', observed=False)['Survived'].value_counts(normalize=True).unstack()
print(age_survival)

# Vẽ biểu đồ
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='AgeGroup', y='Survived', palette='magma', errorbar=None, order=labels)
plt.title('Tỷ lệ sống sót theo Nhóm tuổi')
plt.ylabel('Tỷ lệ sống sót')
plt.xlabel('Nhóm tuổi')
plt.xticks(rotation=45)
plt.show()
```

```python
# Vẽ biểu đồ Histogram với tần suất tương đối
plt.figure(figsize=(10, 6))

# Lọc dữ liệu: Người chết (0) và Người sống (1)
# Sử dụng common_norm=False để mỗi nhóm có tổng diện tích = 1 (tần suất tương đối)
sns.histplot(data=df, x='Age', hue='Survived', kde=True,
             stat='density', common_norm=False, palette='Set1')

plt.title('Phân phối độ tuổi của Người sống sót và Không sống sót (Tần suất tương đối)')
plt.xlabel('Tuổi')
plt.ylabel('Mật độ (Tần suất tương đối)')
plt.show()
```
